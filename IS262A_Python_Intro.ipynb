```python
path="/content/drive/MyDrive/Library_Usage.csv"
import pandas as pd
df=pd.read_csv(path)
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,8))
df["Year Patron Registered"].value_counts().plot(
    kind="pie"
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
plt.figure(figsize=(8,8))
df["Year Patron Registered"].value_counts().plot(
    kind="bar"
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
plt.figure(figsize=(8,8))
df["Year Patron Registered"].value_counts().plot(
    kind="bar"
)
plt.title("Patron Registrations by Year")
plt.xlabel("Year")
plt.ylabel("Number of Registrations")
plt.tight_layout()
plt.show()
```

```python
df["Circulation Active Month"].value_counts().plot(
    kind="bar",
    figsize=(11,8),
    color=["orchid"]

)
plt.title("Library Circulation Activity by Month")
plt.ylabel("Number of Circulation Activity")
plt.xlabel("Month")
plt.tight_layout()
plt.show()
```

```python
month_order=[
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10,6),
    color="aquamarine",
    edgecolor="black"
)
plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month",fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
df["Age Range"].value_counts().plot(
    kind="pie"
)
```

```python
df["Age Range"].value_counts().plot(
    order= "0 to 9 years", "10 to 19 years", "20 to 24 years", "25 to 34 years", "35 to 44 years", "45 to 54 years", "55 to 59 years", "60 to 64 years", "65 to 74 years", "75 years and over",
    kind="bar",
    color="yellowgreen")
```
